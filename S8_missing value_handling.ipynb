# Fill With Methods

- Whenever we are using methods no need to provide values

- pad

- bfill

- ffill

- backfill

```python
data = {'Names':['Ramesh', 'Suresh', 'Satish', np.nan],
       'Age':[np.nan, 22, 25, 27],
       'City': ['Pune', 'Mum', np.nan, 'Pune']}

df = pd.DataFrame(data)
# df.to_csv('missing_nan.csv', index = False)
df
```

```python
df.fillna(method = 'pad') # replaces value by row 
```

```python
import warnings
warnings.filterwarnings('ignore') # used to ignore warnings
df.fillna(method = 'pad') # replaces value by row 
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method = 'pad', axis=1)
```

```python
warnings.filterwarnings('ignore')
df.fillna(method = 'ffill')
```

```python
warnings.filterwarnings('ignore')
df.fillna(method = 'bfill')
```

```python
warnings.filterwarnings('ignore')
df.fillna(method = 'backfill')
```

# Mean - Median - Mode

- Mean >: use mean when there are not outliers data

- Median >: use median when there are outliers

```python
age_mean = df['Age'].mean()
age_mean
```

```python
df[['Age']].fillna(age_mean)
```

```python
age_med = df['Age'].median()
df[['Age']].fillna(age_med)
```

```python
mode = df['City'].mode()
mode # answer coming in series
```

```python
mode = df['City'].mode()
mode.values # we aplied .values so we get dataframe
```

```python
mode = df['City'].mode()
mode.values[0]
```

```python
age_mod = df['City'].mode()
df[['City']].fillna(age_mod)
```
