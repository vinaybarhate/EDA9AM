#**LABELENCODER VS ONEHOTENCODER**
- 

- Generally we dont want a relation ship between input columns

- we need both input columns are not correlated each other

- we need both input columns are **independent(XY)** of each other

- we need both input columns are **perpendicular** to each other

- we need both input columns are **orthogoanal** each other

- we need both input columns are **90 degree phase** shift each other

- OneHotEncoder will achive both

- But OneHotEncoder creates new columns

     - columns increase means dimensions increases

     - more dimensions data need more processing time

     - more dimension data need more infrastructure

     - This is called **curse of dimensionality**
