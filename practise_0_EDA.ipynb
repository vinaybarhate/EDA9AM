**Different Operations on DataFrame**
- a DataFrame for example std_d
- To acess only specific columns std_d[[column, names]] always use double [[]]
- [] it gives output as series not a dataframe. because series is 1d and dataframe is 2d so we should always use double sq bracktes
- you can access multiple columns

- you can also access a specific data point from a column ex : std_d['Marks'][2] you marks of 2 index student

- std_d['Name'][3],  '---------->' ,std_d['Marks'][4]

- You can replace data by using std_d.replace(['chemistry', 'Biology']) But it rplaces all the values with the given value

- so for replacing a specific value you should use loc : std_d.loc[2, 'Marks;]='Psychology' (rows, columns) : you can put index of row but can not put index of column. you can use iloc for that purpose

- you can insert a row using loc but you can not insert a row using iloc : std_d['Grade']=['CD', 'AB', 'BC', 'AB', 'BB']

**Step 12: How to add / append a row**



- we will use loc

```python
num.loc[1:3, 1:2] # we can not insert the index number of columns in loc we have to provide the names of the column instead
```

```python
num.iloc[1:6, 1:3] # in iloc we can directly put index of columns unlike loc
# i in iloc stands for index
```

```python
num.loc[1:6, ['sq', 'cube']]
```

```python
cols= ['sq', 'cube']
num.loc[1:6,cols] # loc doent take index of columns so we have to provide name of the columns
```
